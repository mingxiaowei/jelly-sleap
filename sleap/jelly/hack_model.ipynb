## update labeled instances

```python
# reload the labels 
labels_1 = sleap.load_file(label_dir)
labels_1.skeletons = new_skeletons
labels_1.videos[0].backend.filename = '/home/mingxiao/jellyfish/video/full_video_1.avi'
```

```python
for lf in labels_1.labeled_frames:
    new_instances = [None for _ in range(tb_cnt)]
    pred_inst = None
    for inst in lf.instances: # each frame has at most 2 instances: labeled and predicted
        if isinstance(inst, sleap.PredictedInstance):
            pred_inst = inst
            break
    if pred_inst is not None:
        lf.instances.remove(pred_inst)
    if len(lf.instances) == 0:
        continue
    assert len(lf.instances) == 1, f'should only have one instance in each frame, but got {len(lf.instances)}'
    for old_node, old_point in zip(lf.instances[0].nodes, lf.instances[0].points):
        if old_node.name.lower() == 'mouth':
            continue
        tb_idx = int(old_node.name[2:])
        tb_skeleton = new_skeletons[tb_idx - 1]
        point_dict = {f'tb{tb_idx}_node': sleap.instance.Point(x=old_point.x, y=old_point.y)}
        tb_instance = sleap.Instance(skeleton=tb_skeleton, points=point_dict, frame=lf)
        new_instances[tb_idx - 1] = tb_instance
    lf.instances = new_instances
```
